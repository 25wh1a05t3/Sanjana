```python
import datetime
import random

print("\n--- Task 1: Data Generation and Basic Structures ---")

# Global Data
DEVICE_TYPES = [
    'Temperature_Sensor',
    'Humidity_Sensor',
    'Pressure_Sensor',
    'Light_Sensor'
]

LOCATIONS = [
    'Lab1',
    'Lab2',
    'OfficeA',
    'OfficeB',
    'Warehouse'
]

READING_RANGES = {
    'Temperature_Sensor': (18.0, 30.0),
    'Humidity_Sensor': (30.0, 80.0),
    'Pressure_Sensor': (980.0, 1030.0),
    'Light_Sensor': (100, 1000)
}

# =====================================================
# Task 1.1 - Generate 100 simulated sensor readings
# =====================================================

raw_sensor_data = []

start_time = datetime.datetime.now()

for i in range(100):

    device_type = random.choice(DEVICE_TYPES)
    location = random.choice(LOCATIONS)

    minimum, maximum = READING_RANGES[device_type]

    value = round(random.uniform(minimum, maximum), 2)

    timestamp = (
        start_time + datetime.timedelta(minutes=i)
    ).strftime("%Y-%m-%d %H:%M:%S")

    device_id = f"DEV{i+1:03d}"

    reading = (
        timestamp,
        device_id,
        device_type,
        location,
        value
    )

    raw_sensor_data.append(reading)

print("Total Sensor Readings:", len(raw_sensor_data))

# =====================================================
# Task 1.2 - Create Temperature Reading Dictionary List
# =====================================================

temperature_readings = []

for reading in raw_sensor_data:

    if reading[2] == "Temperature_Sensor":

        temperature_readings.append({

            "timestamp": reading[0],
            "device_id": reading[1],
            "location": reading[3],
            "temperature_celsius": reading[4]

        })

print("Temperature Readings:", len(temperature_readings))

# =====================================================
# Task 1.3 - Create Set of Unique Locations
# =====================================================

unique_locations = set()

for reading in raw_sensor_data:
    unique_locations.add(reading[3])

print("\nUnique Locations")
print(unique_locations)

# =====================================================
# Task 1.4 - Device Statistics Dictionary
# =====================================================

device_stats = {}

for device in DEVICE_TYPES:
    device_stats[device] = (float("inf"), float("-inf"))

for reading in raw_sensor_data:

    device = reading[2]
    value = reading[4]

    minimum, maximum = device_stats[device]

    minimum = min(minimum, value)
    maximum = max(maximum, value)

    device_stats[device] = (minimum, maximum)

print("\nDevice Statistics")

for device, stats in device_stats.items():
    print(f"{device} : Min = {stats[0]}, Max = {stats[1]}")

# =====================================================
# Task 1.5 - Indexing and Slicing
# =====================================================

print("\n--- Task 1.5: Advanced Indexing and Slicing ---")

sample_string = "PythonProgrammingIsFun"

# -------------------------------
# Task 1.5.1 - List Slicing
# -------------------------------

sliced_readings_1 = raw_sensor_data[9:20]

sliced_readings_2 = raw_sensor_data[::5]

print("\n10th to 20th Readings")
for item in sliced_readings_1:
    print(item)

print("\nEvery 5th Reading")
for item in sliced_readings_2:
    print(item)

# -------------------------------
# Task 1.5.2 - String Slicing
# -------------------------------

substring_1 = sample_string[6:17]

substring_2 = sample_string[:6][::-1]

print("\nSubstring 1 :", substring_1)
print("Substring 2 :", substring_2)

# -------------------------------
# Task 1.5.3 - Mixed Slicing
# -------------------------------

fifth_reading_device_type = raw_sensor_data[4][2]

device_type_suffix = fifth_reading_device_type[-6:]

print("\n5th Reading Device Type :", fifth_reading_device_type)
print("Last 6 Characters :", device_type_suffix)

print("\n--- Task 1 Completed Successfully ---")
print("\n--- Task 2: Function Definitions and Usage ---")

# ==========================================================
# Task 2.1
# Function to get readings by location
# ==========================================================

def get_readings_by_location(readings_list, target_location):
    """
    Returns all sensor readings for the given location.
    """
    location_readings = []

    for reading in readings_list:
        if reading[3] == target_location:
            location_readings.append(reading)

    return location_readings


# ==========================================================
# Task 2.2
# Function to calculate average temperature
# ==========================================================

def calculate_average_temperature(temp_data_list):
    """
    Calculates and returns the average temperature.
    Returns 0.0 if no temperature readings exist.
    """
    if len(temp_data_list) == 0:
        return 0.0

    total = 0

    for reading in temp_data_list:
        total += reading["temperature_celsius"]

    average = total / len(temp_data_list)

    return average


# ==========================================================
# Task 2.3
# Function to generate report line
# ==========================================================

def generate_report_line(reading_dict):
    """
    Returns a formatted report line.
    """

    return (
        f"[{reading_dict['timestamp']}] "
        f"Device {reading_dict['device_id']} "
        f"at {reading_dict['location']}: "
        f"{reading_dict['temperature_celsius']}°C"
    )


# ==========================================================
# Testing Task 2 Functions
# ==========================================================

print("\nTesting Task 2 Functions")

# Test Task 2.1
lab1_readings = get_readings_by_location(raw_sensor_data, "Lab1")

print("\nNumber of readings from Lab1:", len(lab1_readings))

# Display first 5 readings
print("\nFirst 5 Lab1 Readings:")

for reading in lab1_readings[:5]:
    print(reading)

# Test Task 2.2
average_temp = calculate_average_temperature(temperature_readings)

print("\nAverage Temperature:", round(average_temp, 2), "°C")

# Test Task 2.3
print("\nSample Report Lines")

for reading in temperature_readings[:5]:
    print(generate_report_line(reading))

print("\n--- Task 2 Completed Successfully ---")
```
