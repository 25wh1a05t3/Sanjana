```python
#EXERCISE-2
import numpy as np

def cosine(u, v):
    return np.dot(u, v) / (np.linalg.norm(u) * np.linalg.norm(v))



pairs = [
    (np.array([1, 0, 1]), np.array([1, 0, 1])),
    (np.array([1, 2, 3]), np.array([3, 2, 1])),
    (np.array([2, 0, 0]), np.array([0, 5, 0]))
]
similarities = []
for i, (u, v) in enumerate(pairs, start=1):

    l1_norm = np.linalg.norm(u, ord=1)
    l2_norm = np.linalg.norm(u, ord=2)

    print("Pair", i)
    print("u =", u)
    print("v =", v)
    print("L1 norm of u:", l1_norm)
    print("L2 norm of u:", l2_norm)

    similarity = cosine(u, v)
    similarities.append(similarity)

    print("Cosine similarity:", similarity)
    print()
most_similar_pair = np.argmax(similarities) + 1

print("Most similar pair =", most_similar_pair)
```

```python
# 3A. CORE MATRIX OPERATIONS
A = np.array([[25., 6.],
              [5., 34.]])
B = np.array([[13., 0.],
              [4., 2.]])

print('A @ B (matrix multiply):\n', A @ B)
print('A.T   (transpose):\n', A.T)
print('inverse(A):\n', la.inv(A))
print('trace(A) = sum of diagonal:', np.trace(A))
```

```python
# 🔹 3B. SPECIAL MATRICES: identity, symmetric, orthogonal
I = np.eye(2)
print('Identity I:\n', I)
print('A @ inv(A) == I ?', np.allclose(A @ la.inv(A), I))
print('A symmetric?', np.allclose(A, A.T))
theta = np.radians(30)
Q = np.array([[np.cos(theta), -np.sin(theta)],
              [np.sin(theta),  np.cos(theta)]])
print('Q orthogonal?', np.allclose(Q.T @ Q, I))
```

```python
#EXERCISE-3
import numpy as np
M = np.array([[4, 2],
              [2, 3]])

P = np.array([[0, -1],
              [1,  0]])
inverse_M = np.linalg.inv(M)
trace_M = np.trace(M)

print("Matrix M:")
print(M)

print("\nInverse of M:")
print(inverse_M)

print("\nTrace of M:", trace_M)
identity = M @ inverse_M
I = np.eye(M.shape[0])

print("\nM @ inverse(M):")
print(identity)

print("\nIs M @ inverse(M) an identity matrix?")
print(np.allclose(identity, I))
is_symmetric = np.allclose(M, M.T)

is_orthogonal = np.allclose(P.T @ P, np.eye(P.shape[0]))

print("\nIs M symmetric?")
print(is_symmetric)

print("\nIs P orthogonal?")
print(is_orthogonal)
```

```python
# 4A. BUILD TRANSFORMATION MATRICES
square = np.array([[0, 1, 1, 0],
                   [0, 0, 1, 1]], dtype=float)
S = np.array([[1.5, 0.0],
              [0.0, 0.5]])

t = np.radians(30)
R = np.array([[np.cos(t), -np.sin(t)],
              [np.sin(t),  np.cos(t)]])

scaled  = S @ square
rotated = R @ square
print('Rotated corners:\n', rotated)
```

```python
# 4B. PLOT ORIGINAL vs TRANSFORMED
def close_loop(pts):

    return np.hstack([pts, pts[:, :1]])

fig, ax = plt.subplots(figsize=(5, 5))
ax.plot(*close_loop(square),  marker='o', label='original')
ax.plot(*close_loop(scaled),  marker='o', label='scaled')
ax.plot(*close_loop(rotated), marker='o', label='rotated 30 deg')
ax.axhline(0, color='gray', lw=0.5); ax.axvline(0, color='gray', lw=0.5)
ax.set_aspect('equal'); ax.legend(); ax.set_title('Linear transformations')
plt.show()
```

```python
#EXERCISE-4
import numpy as np
import matplotlib.pyplot as plt
tri = np.array([[10, 6, 0.5],
                [7, 0, 1.0]])
S = np.array([[6, 0],
              [0, 0.5]])

print("Scaling Matrix:")
print(S)
theta = np.radians(45)

R = np.array([[np.cos(theta), -np.sin(theta)],
              [np.sin(theta),  np.cos(theta)]])

print("\nRotation Matrix:")
print(R)
scaled_tri = S @ tri
transformed_tri = R @ scaled_tri

print("\nOriginal Triangle:")
print(tri)

print("\nScaled Triangle:")
print(scaled_tri)

print("\nTransformed Triangle:")
print(transformed_tri)

original_closed = np.hstack([tri, tri[:, 0:1]])
transformed_closed = np.hstack([transformed_tri,
                                transformed_tri[:, 0:1]])

plt.figure(figsize=(7, 6))

plt.plot(original_closed[0],
         original_closed[1],
         'bo-',
         label='Original Triangle')

plt.plot(transformed_closed[0],
         transformed_closed[1],
         'ro-',
         label='Transformed Triangle')

plt.xlabel("X")
plt.ylabel("Y")
plt.title("Original vs Transformed Triangle")
plt.axhline(0, color='black', linewidth=0.8)
plt.axvline(0, color='black', linewidth=0.8)
plt.grid(True)
plt.legend()
plt.axis('equal')

plt.show()
```

```python
# 5A. COMPUTE EIGENVALUES & EIGENVECTORS
A = np.array([[22., 10.],
              [0., 32.]])

vals, vecs = la.eig(A)
print('Eigenvalues  (lambda):', vals)
print('Eigenvectors (columns):\n', vecs)
```

```python
# 5B. VERIFY  A v = lambda v
for i in range(len(vals)):
    v = vecs[:, i]
    lhs = A @ v
    rhs = vals[i] * v
    print(f'lambda={vals[i]:.1f}  A v == lambda v ?', np.allclose(lhs, rhs))
```

```python
#EXERCISE-5
import numpy as np

C = np.array([[4, 1],
              [2, 3]])

eigenvalues, eigenvectors = np.linalg.eig(C)

print("Eigenvalues:")
print(eigenvalues)

print("\nEigenvectors:")
print(eigenvectors)

print("\nEigenvalues:")
for i, value in enumerate(eigenvalues, start=1):
    print("Eigenvalue", i, "=", value)
lambda_1 = eigenvalues[0]
v_1 = eigenvectors[:, 0]

left_side = C @ v_1
right_side = lambda_1 * v_1

print("\nFirst eigenvalue:", lambda_1)
print("First eigenvector:", v_1)

print("\nC @ v:")
print(left_side)

print("\nlambda * v:")
print(right_side)

print("\nVerification:")
print(np.allclose(left_side, right_side))
```

```python
#6A. SOLVE A 3x3 SYSTEM  A x = b
A = np.array([[ 2.,  1., -1.],
              [-3., -1.,  2.],
              [-2.,  1.,  2.]])
b = np.array([8., -11., -3.])

x = la.solve(A, b)
print('Solution x:', x)
print('Rank of A :', la.matrix_rank(A))
print('Full rank -> a unique solution exists')
```

```python
#6B. RANK TELLS YOU SOLVABILITY
D = np.array([[12., 32., -3.],
              [42., 5., 6.],
              [5., 75., 9.]])
print('Rank of D:', la.matrix_rank(D), '-> < 3, so rows are dependent')
```
